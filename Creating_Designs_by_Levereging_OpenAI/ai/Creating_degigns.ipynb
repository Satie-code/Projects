```python
import gradio as gr
import requests
from PIL import Image
from io import BytesIO
from openai import OpenAI
```

```python
import os
from getpass import getpass

api_key = getpass("Enter your OpenAI API key: ")

os.environ["OPENAI_API_KEY"] = api_key

client = OpenAI(api_key=api_key)

print("OpenAI client configured successfully!")
```

```python
def generate_image(prompt):
    response = client.images.generate(
        model="gpt-image-1",
        prompt=prompt,
        size="1024x1024"
    )

    image_data = response.data[0].b64_json

    image_bytes = BytesIO(__import__("base64").b64decode(image_data))

    image = Image.open(image_bytes)

    return image
```

```python
test_prompt = "A futuristic science-fiction movie poster for a Netflix campaign, cinematic lighting, dramatic atmosphere"

generated_image = generate_image(test_prompt)

generated_image
```

```python
demo = gr.Interface(
    fn=generate_image,
    inputs=gr.Textbox(
        label="Enter your design prompt",
        placeholder="Describe the banner or poster you want to create..."
    ),
    outputs=gr.Image(
        label="Generated Design"
    ),
    title="AI Design Generator",
    description="Generate customized visual designs using OpenAI."
)

print("Gradio interface created successfully!")
```

```python
demo.launch()
```
