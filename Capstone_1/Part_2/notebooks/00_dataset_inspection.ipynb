```python

from pathlib import Path
import pandas as pd
import numpy as np

current = Path.cwd().resolve()

ROOT = next(
    (p for p in [current, *current.parents]
     if p.name == "Capstone_1"),
    None
)

if ROOT is None:
    raise FileNotFoundError(
        "Could not locate the Capstone_1 project directory."
    )

TESLA_CSV = ROOT / "Part_2" / "data" / "raw" / "Tesla_Deaths.csv"

print("Project root:", ROOT)
print("CSV exists:", TESLA_CSV.exists())
```

```python

if not TESLA_CSV.exists():
    raise FileNotFoundError(f"Tesla dataset not found: {TESLA_CSV}")

tesla_df = pd.read_csv(TESLA_CSV)

print("Dataset shape:", tesla_df.shape)

print("\nColumn names:")
print(tesla_df.columns.tolist())

print("\nFirst 10 rows:")
display(tesla_df.head(10))

print("\nData types:")
display(tesla_df.dtypes.to_frame("Data Type"))

print("\nMissing values:")
display(tesla_df.isnull().sum().to_frame("Missing Values"))

print("\nDuplicate rows:", tesla_df.duplicated().sum())

print("\nUnique values per column:")
display(tesla_df.nunique(dropna=True).to_frame("Unique Values"))

print("\nNumerical summary:")
display(tesla_df.describe(include="all").T)
```

```python

print("Potential date-related columns:")

date_columns = [
    column for column in tesla_df.columns
    if "date" in column.lower() or "year" in column.lower()
]

print(date_columns)

print("\nPotential categorical columns:")

categorical_columns = tesla_df.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print(categorical_columns)

for column in categorical_columns:
    print(f"\n{column}:")
    print("Unique values:", tesla_df[column].nunique(dropna=True))
    display(tesla_df[column].value_counts(dropna=False).head(10))

```
